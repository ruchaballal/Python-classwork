```python
#recursion 
#recursion is a technique in which a function calls itself to solve a problem
def count(n):
    if n==0: #base case:case where we need to stop the function
        return #exit this function now, dont return anything
    print (n)
    count(n-1)
count (6)
```

```python
#factorial
def factorial (n):
    if n==0:
        return 1

    return n*factorial (n-1)
factorial (5) #calling my function
```

```python
'''write a python program that accepts the sides of triangle as inputs. the program should indicate whether or not the 
triangle is a right angle triangle using functions'''
h=int (input ("Enter length of 1st side"))
k= int (input ("Enter length of 2nd side"))
l= int(input("Enter length of 3rd side"))
def rightangle (h,k,l):
    if ((h)^2+(k)^2== (l)^2):
        print ("Right angled triangle")
    elif ((k)^2+ (l)^2 == (h)^2):
        print ("Right angled triangle")
    elif ((h)^2+(l)^2== (k)^2):
        print ("Right angled triangle")
    else:
        print ("Not a right angled triangle")
rightangle(h,k,l)
```
