```python
#write an algorithm for the following code
def fact(n):
    if n==0:
        return 1
    else:
        return n*fact(n-1)
```

```python
import re
txt="hello planet"
#check if the string starts with hello
x=re.findall("^hello",txt)
if x:
    print("Yes matched")
else:
    print ("No match")
```

```python
import re
txt="hello planet"
#check if the string ends with planet
x=re.findall("planet$",txt)
if x:
    print ("Yes")
else:
    print("No")
```

```python
#(.)=wild character
import re
text="cat cut cot c9t"
result=re.findall(r"c.t",text)#wild character
result1=re.findall(r"c.*t",text)#wild character with *
result2=re.findall(r"c.+",text)#wild character with +
print(result)
print(result1)
print(result2)
```

```python
import re

text="cat cut cot c9t, ct, cry"
res=re.findall(r"c.+?t",text)#wild character with +
print(res)
```
